```python
import pandas as pd

df = pd.read_csv('test_Y3wMUe5_7gLdaTN(1).csv')

print("First 5 records:")
print(df.head())
```

CHECKING MISSING VALUES

```python
print("\nMissing values in each column:")
missing_data = df.isnull().sum()
print(missing_data)
```

CHECKING DUPLICATE RECORDS

```python
print("\nNumber of duplicate records:")
duplicate_count = df.duplicated().sum()
print(duplicate_count)
```

HANDLING CATEGORICAL MISSING VALUES

```python
print("\nHandling missing categorical values:")

df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
df['Self_Employed'] = df['Self_Employed'].fillna(
    df['Self_Employed'].mode()[0]
)

print("Categorical missing values handled successfully.")
```

HANDLING NUMERICAL MISSING VALUES

```python
print("\nHandling missing numerical values:")

df['LoanAmount'] = df['LoanAmount'].fillna(
    df['LoanAmount'].median()
)

df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(
    df['Loan_Amount_Term'].mode()[0]
)

print("Numerical missing values handled successfully.")
```

CHECKING CLEANED DATA

```python
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```
